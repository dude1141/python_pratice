```python
for i in range(5):
    for j in range(i+1):
        print("*", end="")
    print()

#outer loop Rows
#inner loop columns , but you need * in sameline you have to use end=""
```

```python
# here start is 5, stop at 0 and come from back like decrement -1

for i in range(5,0,-1):
    for j in range(i+1):
        print("*", end="")
    print()
```
